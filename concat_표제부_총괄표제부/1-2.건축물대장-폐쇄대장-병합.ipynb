# 파일 읽기 및 전처리

```python
blds_path = "../interim/blds.csv"
closed_blds_path = "../input/closed_blds.csv"
```

```python
blds = pd.read_csv(blds_path)           # 건축물대장에는 이전 단계에서 이미 병합종류 컬럼을 추가하므로 따로 컬럼 추가하지 않음

closed_blds = pd.read_csv(closed_blds_path)
closed_blds.insert(0, '병합종류', '0번')    # 폐쇄대장에 병합종류 컬럼 추가
```

```python
closed_blds = closed_blds.drop('동명칭')    # ['건물명'] 컬럼과 같은 의미의 컬럼이므로 드랍

# 폐쇄대장을 건축물대장과 동일한 형식으로 수정
closed_blds = closed_blds.rename(columns=corresponding_columns)
closed_blds = closed_blds.reindex(columns=blds.columns.tolist() + ['폐쇄일'])   
```

# 병합

```python
result = pd.concat([blds,closed_blds], ignore_index=True)
```

# 결과

```python
result.to_csv("../interim/blds.csv", index=False)
```
