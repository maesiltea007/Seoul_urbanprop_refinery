건축물대장(1번)이 포함되지 않은 row는 전부 드랍하여 별개의 엑셀파일에 보관한다.

```python
import pandas as pd
```

# 파일 읽기

```python
csv_path = "../input/final.csv"

df = pd.read_csv(csv_path)
```

# 분리

```python
successed_df = df[df["병합종류"].str.startswith("1번")]
failed_df = df[~df["병합종류"].str.startswith("1번")]
```

# 디자인 추가

```python
color_map = {
    'kapt_id': 'background-color: red',
    'sapt_id': 'background-color: orange',
    'blds_id': 'background-color: lightgreen',
    'unts_summary_id': 'background-color: lightblue'
}

def color_columns(df):
    styles = pd.DataFrame('', index=df.index, columns=df.columns)
    for col, style in color_map.items():
        if col in df.columns:
            styles[col] = style
    return styles
```

```python
styled_successed_df = successed_df.copy()
styled_failed_df = failed_df.copy()

styled_successed_df = styled_successed_df.style.apply(color_columns, axis=None)
styled_failed_df = styled_failed_df.style.apply(color_columns, axis=None)
```

# 결과

```python
styled_successed_df.to_excel("../output/1번_포함.xlsx", index=False)
styled_failed_df.to_excel("../output/1번_없는_것.xlsx", index=False)
```
